```python
import jsonimport pandas as pdimport matplotlib.pyplot as plt
```

```python
with open('gold_prices.json', 'r') as f:    data = json.load(f)
```

```python
df = pd.DataFrame(data)df['Date_'] = pd.to_datetime(df['Date_'])df.set_index('Date_', inplace=True)
```

```python
plt.figure(figsize=(14, 7))plt.plot(df.index, df['Close_GC=F'])plt.title('Gold Prices Over Time')plt.xlabel('Date')plt.ylabel('Price (USD)')plt.grid(True)plt.show()
```

```python
df['SMA_50'] = df['Close_GC=F'].rolling(window=50).mean()df['SMA_200'] = df['Close_GC=F'].rolling(window=200).mean()
```

```python
plt.figure(figsize=(14, 7))plt.plot(df.index, df['Close_GC=F'], label='Close Price')plt.plot(df.index, df['SMA_50'], label='50-Day SMA')plt.plot(df.index, df['SMA_200'], label='200-Day SMA')plt.title('Gold Prices with 50-Day and 200-Day SMAs')plt.xlabel('Date')plt.ylabel('Price (USD)')plt.legend()plt.grid(True)plt.show()
```
